1.What is Inferential Statistics?

Inferential statistics is the branch of statistics used to make conclusions or predictions about a population based on sample data. It helps researchers analyze patterns, relationships, and differences using probability theory.

2.What is Hypothesis Testing and its Components?

Hypothesis testing is a statistical method used to determine whether there is enough evidence to support a claim about a population.

Components:
Null Hypothesis (H₀): Assumes no effect or relationship exists.
Alternative Hypothesis (H₁): Assumes an effect or relationship exists.
Significance Level (α): Probability of rejecting H₀ incorrectly (commonly 0.05).
Test Statistic: Numerical value calculated from sample data.
p-value: Probability of obtaining the observed result if H₀ is true.
3. Explain Confidence Interval and Critical Value
Confidence Interval (CI)

A confidence interval is a range of values likely to contain the true population parameter.

Example:
A 95% CI means we are 95% confident that the true mean lies within the interval.

Critical Value

A critical value is the cutoff point used to decide whether to reject the null hypothesis.

For a 95% confidence level:

z-critical value ≈ 1.96

4.Define p-value

The p-value measures the probability of obtaining the observed results assuming the null hypothesis is true.

If p-value < 0.05 → Reject H₀
If p-value > 0.05 → Fail to reject H₀

5.Differentiate Type I and Type II Errors
Error Type	Meaning
Type I Error	Rejecting a true null hypothesis
Type II Error	Failing to reject a false null hypothesis

6.Brief Description of Statistical Tests
z-test

Used when:

Sample size is large (n > 30)
Population variance is known
t-test

Used when:

Sample size is small
Population variance is unknown
Chi-Square Test

Used to test relationships between categorical variables.

ANOVA

Used to compare means among three or more groups.

7.What is Covariance?

Covariance measures how two variables change together.

Positive covariance → variables increase together
Negative covariance → one increases while the other decreases

8.What is Correlation?

Correlation measures the strength and direction of the relationship between two variables.

+1 → Perfect positive relationship
-1 → Perfect negative relationship
0 → No relationship


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/health_records_dataset.csv")
df.head()
```

```python
from scipy import stats
from statsmodels.stats.weightstats import zconfint
from scipy.stats import ttest_ind, chi2_contingency, f_oneway
```

```python
#Confidence Level
mean_age = df['age'].mean()
std_age = df['age'].std()
n = len(df)

confidence = 0.95
z = 1.96

margin_error = z * (std_age / np.sqrt(n))

lower = mean_age - margin_error
upper = mean_age + margin_error

print("Confidence Interval:", (lower, upper))
```

```python

print("95% Confidence Interval")

numerical_cols = [
    'weight',
    'bmi',
    'blood_pressure',
    'cholesterol_level',
    'glucose_level'
]

for col in numerical_cols:
    ci = zconfint(df[col])
    print(f"{col}: {ci}")
```

```python
from statsmodels.stats.weightstats import ztest

smoker_glucose = df[df['smoking_status'] == 'Smoker']['glucose_level']
nonsmoker_glucose = df[df['smoking_status'] == 'Non-Smoker']['glucose_level']

z_stat, p_value = ztest(smoker_glucose, nonsmoker_glucose)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

```python
# Critical Value and p-value Example
alpha = 0.05
critical_value = stats.norm.ppf(1 - alpha/2)
print("\nCritical Value:", critical_value)
print("Significance Level:", alpha)
```

```python
# t-test
#Compare BMI of Smokers vs Non-Smokers
smokers = df[df['smoking_status'] == 'Smoker']['bmi']
non_smokers = df[df['smoking_status'] == 'Non-Smoker']['bmi']

t_stat, p_value = ttest_ind(smokers, non_smokers)

print("\nT-Test Result")
print("T-Statistic:", t_stat)
print("P-Value:", p_value)

if p_value < 0.05:
    print("Reject H0")
else:
    print("Accept H0")
```

```python
#Chi-Square Test
#Smoking Status vs Diabetes
contingency_table = pd.crosstab(df['smoking_status'], df['diabetes'])

chi2, p, dof, expected = chi2_contingency(contingency_table)

print("\nChi-Square Test")
print("Chi-Square Value:", chi2)
print("P-Value:", p)

if p < 0.05:
    print("Reject H0")
else:
    print("Accept H0")
```

```python
#ANOVA Test
#BMI Across Age Groups

group1 = df[df['age_group'] == '18-25']['bmi']
group2 = df[df['age_group'] == '26-35']['bmi']
group3 = df[df['age_group'] == '36-45']['bmi']
group4 = df[df['age_group'] == '46-60']['bmi']
group5 = df[df['age_group'] == '60+']['bmi']

anova_result = f_oneway(group1, group2, group3, group4, group5)

print("\nANOVA Test")
print("F-Statistic:", anova_result.statistic)
print("P-Value:", anova_result.pvalue)

if anova_result.pvalue < 0.05:
    print("Reject H0")
else:
    print("Accept H0")
```

```python
#Covariance and Correlation

covariance = np.cov(df['age'], df['bmi'])[0][1]
correlation = np.corrcoef(df['age'], df['bmi'])[0][1]

print("\nCovariance:", covariance)
print("Correlation:", correlation)
```
